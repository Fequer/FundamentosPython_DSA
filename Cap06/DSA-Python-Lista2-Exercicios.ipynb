#### Exercício 1

Escreva uma função que receba os valores de três lados de um triângulo e o classifique como "Equilátero" (todos os lados iguais), "Isósceles" (dois lados iguais) ou "Escaleno" (todos os lados diferentes).

```python
#### Solução

lado_a = input("Digite o primeiro lado do triângulo: ")
lado_b = input("Digite o segundo lado do triângulo: ")
lado_c = input("Digite o terceiro lado do triângulo: ")

def classifica_triangulo():
    if lado_a == lado_b and lado_b == lado_c:
        print("Triângulo Equilátero")
    elif lado_a == lado_b or lado_a == lado_c or lado_b == lado_c:
        print("Triângulo Isósceles")
    else:
        print("Triângulo Escaleno")

classifica_triangulo()
```

#### Exercício 2

Escreva uma função que recebe um número inteiro e exibe a tabuada de multiplicação desse número, do 1 ao 10.

```python
# Solução

def tabuada():
    numero = int(input("Digite um número"))
    for n in range(1,11):
        tabuada_resultado = numero * n
        print(f"{numero} X {n} = {tabuada_resultado}")
        
tabuada()
```

#### Exercício 3 

Você recebeu um dicionário com os nomes dos alunos e suas respectivas notas. Escreva uma função que calcula a média da turma e retorna uma lista com os nomes dos alunos que tiveram nota acima da média.

```python
# Solução

alunos = {
    "Felipe": 7, 
    "Jessica": 5, 
    "Saulo": 2, 
    "Bruce": 8
}

def calculo_media():    

    total = sum(alunos.values())
    quantidade_aluno = len(alunos)
    media = (total/quantidade_aluno)
    print(f"A média da turma é {media:.2f}")
    for aluno,nota in alunos.items():
        if nota <= media:
            print(f"{aluno} abaixo da média")
        else:
            print(f"{aluno} acima da média")

calculo_media()
```

#### Exercício 4

Dada uma lista de números, crie uma nova lista usando list comprehension que contenha o quadrado de cada número par da lista original.

```python
# Solução

numeros = [2,3,4,6,8,9,13,20,26,42]
nova_lista = [ x ** 2 for x in numeros if x % 2 == 0]

print(nova_lista)
```

#### Exercício 5

Crie uma função chamada dsa_calcula_imc que aceite dois argumentos: peso (em kg) e altura (em metros). A função deve calcular o Índice de Massa Corporal (IMC) usando a fórmula IMC=peso/altura^2 e retornar o valor do IMC.

```python
# Solução

peso = float(input("Digite o peso (00 kg): "))
altura = float(input("Digite a altura (0.00 m): "))

def dsa_calcula_imc(peso, altura): 
    imc = peso / (altura ** 2)
    return imc

print(f"IMC: {dsa_calcula_imc(peso, altura):.2f}")
```

#### Exercício 6

Você tem uma lista de dicionários, onde cada dicionário representa uma pessoa com nome e idade. Use a função sorted() com uma expressão lambda como chave (key) para ordenar a lista de pessoas da mais nova para a mais velha.

```python
# Solução

pessoas = [
    {"nome":"Felipe", "idade": 33},
    {"nome":"Jessica", "idade": 35},
    {"nome":"Bruce", "idade": 5},
    {"nome":"Saulo", "idade": 3},
]

ordenar = sorted(pessoas, key=lambda x: x["idade"])

print("Ordenada por idade: ", ordenar)
```

#### Exercício 7

Crie uma função que receba uma lista de números inteiros e retorne um dicionário contendo a contagem de quantos números são pares e quantos são ímpares.

```python
# Solução

numeros = [1,5,3,15,32,47,95,42,142,36,547,791]

def par_ou_impar(numeros):
    resultado = {
        "pares": 0,
        "impares": 0
    }
    for numero in numeros:
        if numero % 2 == 0:
            resultado["pares"] = resultado["pares"] + 1
        else:
            resultado["impares"] +=1
    return resultado

def contar_pares_impares(numeros):
    return {
        "Pares": sum(1 for numero in numeros if numero % 2 == 0),
        "Impares": sum(1 for numero in numeros if numero % 2 != 0)
    }

print(par_ou_impar(numeros))
print(contar_pares_impares(numeros))
```

#### Exercício 9

Dada uma lista de frases, use a função map() em conjunto com uma expressão lambda para criar uma nova lista onde cada frase é convertida para letras maiúsculas e tem a palavra "PYTHON" anexada ao final.

```python
# Solução

frases = ["Estudando programação pela dsa ", 
          "programação é muito importante ", 
          "resolvendo problemas com computação ", 
          "utilizando jupiternotebook "
         ]

nova_frase = list(map(lambda frase: frase.upper() + "PYTHON", frases))

print(nova_frase)
```

```python
#### Solução
import random

print( 20 * "-", "Adivinhe o número", 20 * "-", "\n", 20 * "-","você tem apenas 5 tentativas", 20 * "-")

def gerar_numero():
    return random.randint(1,20)
    

numero_gerado = gerar_numero()
contador = 5

print(numero_gerado)

escolha = int(input("Digite o número entre 1 e 20: "))

def verificar(escolha, contador, numero_gerado):
        while contador > 0:
            if escolha == numero_gerado:
                print (25 * "-", "Você acertou !", 25 * "-")
                break
            else:
                contador -= 1
                if contador == 0:
                    print(25 * "-","Você errou", 25 * "-","\n\nO número era: ", numero_gerado)
                    break
                else:
                    if escolha > numero_gerado:
                        print(25 * "-","Você errou, o número é menor !", 25 * "-","\n\nTentativas restantes", contador)
                        escolha = int(input("\nTente denovo: "))
                    else:
                        print(25 * "-","Você errou, o número é maior !", 25 * "-","\n\nTentativas restantes", contador)
                        escolha = int(input("\nTente denovo: "))

verificar(escolha, contador, numero_gerado)
```
